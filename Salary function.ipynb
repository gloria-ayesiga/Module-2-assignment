```python

import pandas as pd
import os
import zipfile
os.getcwd()
```

```python
df = pd.read_csv('Total.csv', low_memory=False)  #importing the csv file
file_size = os.path.getsize("Total.csv")
df.head()  #confirming that the headings have been imnported correctly
```

```python
def employee_detail(name, df):  #creating the employee-detail function
   try:
    
    result = df[df['EmployeeName'].str.contains(name, case=False, na = False)]
    
    if result.empty:
        return f"No employee found with the name: {name}"
    else:
        return result
   except KeyError as e:
        return f"column not found: {e}"
   except Exception as e:
        return "f error occured: {e}"   

employee_detail("christopher chong", df) #testing out the function's output
```

```python
#cleaning data block code using python dictionary

try:
    data_proc = {}

    # Original DataFrame assumed already loaded into `df`
    data_proc["original"] = df.copy()

    # Clean 'EmployeeName' and 'JobTitle' columns
    data_proc["cleaned"] = data_proc["original"].copy()
    data_proc["cleaned"][["EmployeeName", "JobTitle"]] = data_proc["cleaned"][["EmployeeName", "JobTitle"]].apply(
        lambda col: col.str.title()
    )

    # Identify duplicates
    data_proc["duplicates"] = data_proc["cleaned"][
        data_proc["cleaned"].duplicated(subset=["EmployeeName", "Year"], keep=False)
    ]
    data_proc["num_duplicates"] = data_proc["duplicates"].shape[0]

    print(f"Number of duplicate rows: {data_proc['num_duplicates']}")
    if data_proc["num_duplicates"] == 0:
        print("The 'EmployeeName' column is unique (per year).")
    else:
        print(f"There are {data_proc['num_duplicates']} duplicate rows based on EmployeeName and Year.")

    # Drop duplicates
    data_proc["duped"] = data_proc["cleaned"].drop_duplicates(subset=["EmployeeName", "Year"], keep='first')

    # Sort final cleaned DataFrame
    data_proc["final"] = data_proc["duped"].sort_values(by='EmployeeName')

    # Save cleaned data
    data_proc["output_file"] = "cleaned_Total.csv"
    data_proc["final"].to_csv(data_proc["output_file"], index=False)

except KeyError as e:
    print(f"Column not found: {e}")
except Exception as e:
    print(f"Error occurred during data cleaning: {e}")
```

```python
#exporting an employees detail


emp_data = employee_detail("A Bernard Fatooh", df)

if isinstance(emp_data, pd.DataFrame):
    file = "A Bernard Fatooh"
    csv_filename = f"{file}.csv"
    zip_filename = f"{file}.zip"

    emp_data.to_csv(csv_filename, index=False) #saving the employees data to a CSV file

    with zipfile.ZipFile(zip_filename, 'w', zipfile.ZIP_DEFLATED) as zipf:
        zipf.write(csv_filename) 
    
    os.remove(csv_filename) #removing the csv

    print("Employee details exported successfully.")
else:
    print(emp_data)  # This will show the "not found" message


```
